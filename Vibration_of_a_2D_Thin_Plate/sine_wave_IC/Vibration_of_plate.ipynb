```python
## Import Library
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import numpy as np

## Plate dimensions in Meters
X=round(np.pi,2)
Y=round(np.pi,2)
## Simulations Parameters
#mesh size
nx=int(100*X )+1
ny=int(100*Y) +1  
del_x=X/(100*X)
del_y=Y/(100*Y)
print(f'nx>>{nx}')
print(f'ny>>{ny}')
del_m=del_x
#material properties
E=3e+4
nu=0.3
G=E/((1-nu)*2)
Rho=11000
#speed of wave in medium
ws=(G/Rho)**0.5
# Simulation time in S
c=0.15
delt= round(c*del_m/ws,4) 
print(delt)
tot_time=20
rx= round(((c*delt)/del_x)**2,6) 
ry= round(((c*delt)/del_y)**2,6)
print(rx)    
print(ry)

## grid_points
x=np.linspace(0,X,nx)
y=np.linspace(0,Y,ny)
a,b=np.meshgrid(x,y)

## Initial Condition
Z=np.sin(a)/2
Z[:,-1]=0
Z_previ=Z
Z_crnt=Z_previ
Z_next=np.zeros([315,315])
Z_displacement=[]
## Plot Initial Condition
fig1=plt.figure(figsize=(5,5))
ax=fig1.add_subplot(111, projection="3d")
surf=ax.plot_surface(a,b,Z,cmap="coolwarm")
fig1.colorbar(surf,ax=ax,label="Displacement in m",shrink=0.4,pad=0.1)

plt.title("Initial Displacement")
plt.tight_layout()
plt.show()

# f1=open("z_displacement.txt", "w" )
# for i in range(len(Z))
print(Z)
```

```python
## Solve the Equation with Central Difference Method
file = open ("z_displacement.txt","w")
for t in range(0,int(tot_time/delt)):
    for i in range(0,nx):
        for j in range(1,ny-1):
            if i==0:
                Z_disp= 2*Z_crnt[i,j] - Z_previ[i,j] + (rx)*(Z_crnt[i+1,j] - 2*Z_crnt[i,j]) + (ry) * (Z_crnt[i,j+1] - 2*Z_crnt[i,j] + Z_crnt[i,j-1])
                Z_next[i][j]=Z_disp

            elif i==nx-1:
                Z_disp= 2*Z_crnt[i,j] - Z_previ[i,j] + (rx)*(- 2*Z_crnt[i,j] + Z_crnt[i-1,j]) + (ry) * (Z_crnt[i,j+1] - 2*Z_crnt[i,j] + Z_crnt[i,j-1])
                Z_next[i][j]=Z_disp

            else:
                Z_disp= 2*Z_crnt[i,j] - Z_previ[i,j] + (rx)*(Z_crnt[i+1,j] - 2*Z_crnt[i,j] + Z_crnt[i-1,j]) + (ry) * (Z_crnt[i,j+1] - 2*Z_crnt[i,j] + Z_crnt[i,j-1])
                Z_next[i][j]=Z_disp

    Z_previ= Z_crnt.copy()
    Z_crnt=Z_next.copy()
    Z_displacement.append(Z_crnt)
        
    file.write('{} >> {}\n'.format(t,Z_previ.copy()))
```

```python
## Create GIF Animation:
zmin = Z.min()
zmax = Z.max()
zlim_margin = (zmax - zmin) * 0.1

fig=plt.figure(figsize=(5,5))
ax=fig.add_subplot(111,projection="3d")
ax.view_init(azim=25)
ax.plot_surface(a,b,Z_displacement[0], cmap="coolwarm")
fig.colorbar(surf,ax=ax,label="Displacement in m",shrink=0.4,pad=0.1)
ax.set_zlim(zmin - zlim_margin, zmax + zlim_margin)
ax.set_title("Vibration of Plate")
ax.set_xlabel("X in m")
ax.set_ylabel("Y in m")
ax.set_zlabel("Z in m")
plt.show()
#print(Z_displacement[0:10])
def animate(i):
    ax.clear()
    ax.set_xlim(0, np.pi)
    ax.set_ylim(0, np.pi)
    ax.set_zlim(zmin - zlim_margin, zmax + zlim_margin)
    ax.plot_surface(a,b,Z_displacement[i], cmap="coolwarm")

anim=animation.FuncAnimation(fig,animate,frames=range(0,int(tot_time/delt),10))
writer=animation.FFMpegWriter(fps=(30))
anim.save("Vibration_of_Plate_sinewave.mp4",writer=writer)
```
